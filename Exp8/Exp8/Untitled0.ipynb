```python
import nltk
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet

# -------------------------------------------------
# Download resources (only first time)
# -------------------------------------------------
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt')
nltk.download('punkt_tab')

# -------------------------------------------------
# Function for WSD using Lesk Algorithm
# -------------------------------------------------

def word_sense_disambiguation(sentence, target_word):

    tokens = word_tokenize(sentence)

    sense = lesk(tokens, target_word)

    if sense:
        print("\nTarget Word:", target_word)
        print("Predicted Sense:", sense.name())

        print("\nMeaning:")
        print(sense.definition())

        print("\nExample Usage:")
        print(sense.examples())

    else:
        print("No sense found")


# -------------------------------------------------
# Input
# -------------------------------------------------

sentence = input("Enter sentence: ")
target = input("Enter ambiguous word: ")

word_sense_disambiguation(sentence, target)
```
